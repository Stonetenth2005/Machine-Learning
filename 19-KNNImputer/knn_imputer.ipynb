```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.impute import KNNImputer, SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
df = pd.read_csv('/content/train.csv', usecols=['Age', 'Pclass', 'Fare', 'Survived'])
```

```python
df.isnull().mean()
```

```python
X = df.drop('Survived', axis=1)
y = df['Survived']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=2)
```

```python
knn = KNNImputer()

X_train_imputed = knn.fit_transform(X_train)
X_test_imputed = knn.transform(X_test)
```

```python
lr = LogisticRegression()
lr.fit(X_train_imputed, y_train)

y_pred = lr.predict(X_test_imputed)
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)
```

# Comparing with mean simple imputer

```python
si = SimpleImputer(strategy='mean')

X_train_imputed2 = si.fit_transform(X_train)
X_test_imputed2 = si.transform(X_test)
```

```python
lr = LogisticRegression()
lr.fit(X_train_imputed2, y_train)

y_pred = lr.predict(X_test_imputed2)
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)
```
