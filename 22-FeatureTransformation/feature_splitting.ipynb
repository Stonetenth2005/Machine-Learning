```python
import pandas as pd
```

```python
df = pd.read_csv('/content/train (2).csv')
```

```python
df['Name']
```

```python
df['Title'] = df['Name'].str.split(',', expand=True)[1].str.split('.', expand=True)[0]
```

```python
df[['Title', 'Name']]
```

```python
df.groupby('Title')['Survived'].mean().sort_values(ascending=False)
```

```python
df['Is_married'] = 0
df['Is_married'].loc[df['Title'] == 'Mrs'] = 1
```
