```python
import pandas as pd
```

```python
df = pd.read_csv('/content/train (2).csv')[['Age', 'Pclass', 'SibSp', 'Parch', 'Survived']]
```

```python
df.head()
```

```python
df.dropna(inplace=True)
```

```python
X = df.iloc[:,0:4]
y = df.iloc[:,-1]
```

```python
import numpy as np
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import cross_val_score

np.mean(cross_val_score(LogisticRegression(), X, y, scoring='accuracy', cv=10))
```

# Applying Feature Construction

```python
X['Family_size'] = X['SibSp'] + X['Parch'] + 1
X.head()
```

  Making a family type column, 0 -> alone, 1 -> small, 2 -> big

```python
def famType(num):
  if(num == 1):
    return 0
  elif(num > 1 and num <= 4):
    return 1
  else:
    return 2
```

```python
X['Family_type'] = X['Family_size'].apply(famType)
X.head()
```

```python
X.drop(columns=['SibSp', 'Parch', 'Family_size'], inplace=True)
X.head()
```

```python
np.mean(cross_val_score(LogisticRegression(), X, y, scoring='accuracy', cv=10))
```

Notice the increase in crossval score
